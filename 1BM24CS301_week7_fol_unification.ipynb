```python
def is_variable(x):
    return isinstance(x, str) and x in {"x", "y", "z", "X", "Y", "Z"}


def apply_substitution(term, subs):
    while is_variable(term) and term in subs:
        term = subs[term]
    if isinstance(term, tuple):
        return (term[0],) + tuple(
            apply_substitution(arg, subs) for arg in term[1:]
        )
    return term


def unify_terms(a, b, subs):
    a = apply_substitution(a, subs)
    b = apply_substitution(b, subs)

    if a == b:
        return True

    if is_variable(a):
        subs[a] = b
        return True

    if is_variable(b):
        subs[b] = a
        return True

    if isinstance(a, tuple) and isinstance(b, tuple):
        if a[0] != b[0] or len(a) != len(b):
            return False

        for x, y in zip(a[1:], b[1:]):
            if not unify_terms(x, y, subs):
                return False
        return True

    return False


def unify(p1, args1, p2, args2):

    # Condition 1: Predicate names must be same
    if p1 != p2:
        print("Unification Failed: Predicate names are different")
        return

    # Condition 2: Number of arguments must be same
    if len(args1) != len(args2):
        print("Unification Failed: Different number of arguments")
        return

    subs = {}

    # Condition 3: Check argument compatibility
    for a, b in zip(args1, args2):
        if not unify_terms(a, b, subs):
            print("Unification Failed: Conflicting substitutions")
            return

    print("Unification Successful!")
    print("Substitutions:")

    for var, value in subs.items():
        print(var, "=", apply_substitution(value, subs))


# Example from the PDF
unify("Eats", ["x", "Apple"],
      "Eats", ["Riya", "y"])
```
